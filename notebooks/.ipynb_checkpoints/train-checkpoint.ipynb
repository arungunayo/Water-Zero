If the information is contained in a page requiring navigation, a search form, a filter, or browser interaction, use the relevant Web Agent or Browser capability rather than assuming that a simple page fetch will work.# Hirakud reservoir – weekly forecasting (`train.ipynb`)
Predicts the **change in `pct_full`** 1, 2 and 4 weeks ahead (7, 14, 28 days) and a low-storage warning.

Why weekly: the cleaned data is daily in 2013-2020 but effectively weekly before (forward-filled, ~70 % stale in 2010-12) and after (2023-25). Modelling at one weekly resolution keeps every feature meaning the same thing in every year.

Pipeline: config → load → weekly → weather → features → targets → time split → baselines → models → evaluation → warnings → final fit/save → sanity checks.

Modelling notes
- Target = `pct_full(t+h) − pct_full(t)` (percentage points), so errors are directly comparable with the baselines.
- **No look-ahead:** the origin week must be actually observed; lagged values may use a single-week gap filled from the past only. Gap-interpolated values are used only for *targets*. A test at the end proves the features are causal.
- Hyper-parameters are chosen with `TimeSeriesSplit` on training years only; the weather variant is also chosen on CV, never on the test set.

## 1. Config

```python
import os
os.environ.setdefault("OMP_NUM_THREADS", "1")   # tiny data: many OpenMP threads make gradient boosting far slower
from pathlib import Path
import json, warnings, itertools, time
import numpy as np
import pandas as pd
import requests
import joblib
import matplotlib.pyplot as plt
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.linear_model import Ridge
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.model_selection import TimeSeriesSplit

warnings.filterwarnings("ignore")

ROOT        = Path("..")
CLEAN_CSV   = ROOT / "data/processed/hirakud_clean.csv"
WEATHER_DIR = ROOT / "data/weather"
OUT_DIR     = ROOT / "outputs"
for d in (WEATHER_DIR, OUT_DIR):
    d.mkdir(parents=True, exist_ok=True)

CUTOFF          = pd.Timestamp("2022-01-01")
HORIZONS        = (1, 2, 4)                 # weeks  -> 7, 14, 28 days
WARN_PERCENTILE = 20                        # low-storage threshold, from training data only
SEED            = 42
VERSION         = "0.1.0"
USE_LAG52       = False                     # optional feature: pct_full 52 weeks ago
BACKTEST_YEARS  = range(2016, 2026)
MONSOON_MONTHS  = [6, 7, 8, 9, 10]

# Approximate upstream points of the Hirakud (Mahanadi) catchment, averaged for the 'upstream' variants
UPSTREAM = {
    "raipur":    (21.25, 81.63),   # Mahanadi/Seonath, Chhattisgarh plains
    "sihawa":    (20.55, 81.57),   # Mahanadi headwaters
    "bilaspur":  (22.08, 82.15),   # Seonath/Hasdeo side
    "raigarh":   (21.90, 83.40),   # Ib / Mand catchment, just upstream of the dam
}

GRID = {"max_depth": [3, 4, 6], "learning_rate": [0.03, 0.05, 0.1], "min_samples_leaf": [10, 20]}
np.random.seed(SEED)
```

## 2. Load the cleaned data

```python
def load_clean(path=CLEAN_CSV):
    df = pd.read_csv(path, parse_dates=["date"])
    assert df["date"].is_monotonic_increasing, "dates not sorted"
    assert df["date"].is_unique, "duplicate dates"
    df["stale_value"] = df["stale_value"].astype(bool)
    # stale (carried-forward) readings carry no new information -> blank them. pct_full is NOT clipped at 100.
    df.loc[df["stale_value"], "pct_full"] = np.nan
    return df

clean = load_clean()
DAM_LAT, DAM_LON = clean["lat"].iloc[0], clean["lon"].iloc[0]
IMPLIED_CAPACITY = (clean["storage_bcm"] / clean["pct_full"] * 100).median()
print(clean.shape, clean.date.min().date(), "->", clean.date.max().date())
print(f"Stale rows blanked: {clean.pct_full.isna().sum()} | implied capacity ~ {IMPLIED_CAPACITY:.3f} bcm")
```

## 3. Weekly resolution (weeks end on Thursday, as in the CWC bulletin)

```python
def to_weekly(df, verbose=False):
    """Weekly mean of non-stale pct_full on W-THU.
    pct_obs  : observed weeks only (NaN where no real reading)
    pct_fill : pct_obs with *single-week* gaps linearly interpolated (longer gaps stay NaN)"""
    s = df.set_index("date")["pct_full"].dropna()
    pct_obs = s.resample("W-THU").mean()
    isna = pct_obs.isna()
    run_id = (isna != isna.shift()).cumsum()
    gap_len = isna.groupby(run_id).transform("sum").where(isna, 0)      # length of the gap each NaN belongs to
    interp = pct_obs.interpolate(limit_area="inside")
    pct_fill = pct_obs.copy()
    fill_mask = isna & (gap_len == 1) & interp.notna()
    pct_fill[fill_mask] = interp[fill_mask]
    w = pd.DataFrame({"pct_obs": pct_obs, "pct_fill": pct_fill})
    if verbose:
        print(f"weeks: {len(w)} | observed: {w.pct_obs.notna().mean():.1%} | after single-gap fill: {w.pct_fill.notna().mean():.1%}"
              f" | usable weeks: {w.pct_fill.notna().sum()}")
    return w

weekly = to_weekly(clean, verbose=True)
weekly.tail()
```

## 4. Weather (Open-Meteo archive, cached to `data/weather/`)
Daily rain, ET0 and temperature for the dam and for the upstream points. Rolling windows **end on the week-end date** so nothing from the future is used.

```python
def fetch_weather(name, lat, lon, start, end):
    f = WEATHER_DIR / f"{name}.csv"
    if f.exists():
        return pd.read_csv(f, parse_dates=["date"]).set_index("date")
    r = requests.get("https://archive-api.open-meteo.com/v1/archive", timeout=60, params=dict(
        latitude=lat, longitude=lon, start_date=start, end_date=end, timezone="Asia/Kolkata",
        daily="precipitation_sum,et0_fao_evapotranspiration,temperature_2m_mean"))
    r.raise_for_status()
    d = pd.DataFrame(r.json()["daily"]).rename(columns={
        "time": "date", "precipitation_sum": "rain", "et0_fao_evapotranspiration": "et0", "temperature_2m_mean": "temp"})
    d["date"] = pd.to_datetime(d["date"])
    d.to_csv(f, index=False)
    time.sleep(1)
    return d.set_index("date")

W_START = (clean.date.min() - pd.Timedelta(days=120)).strftime("%Y-%m-%d")     # warm-up for 13-week windows
W_END   = clean.date.max().strftime("%Y-%m-%d")
points  = {"dam": (DAM_LAT, DAM_LON), **UPSTREAM}
wx_raw  = {n: fetch_weather(n, la, lo, W_START, W_END) for n, (la, lo) in points.items()}
for n, d in wx_raw.items():
    print(f"{n:9s} rows={len(d)} missing={int(d.isna().sum().sum())}")
```

```python
def add_weather(wx_daily):
    """Daily weather -> features on W-THU week ends (windows end on the week-end date)."""
    d = wx_daily.asfreq("D")
    d[["rain", "et0", "temp"]] = d[["rain", "et0", "temp"]].interpolate(limit=3)
    f = pd.DataFrame({
        "rain_1w":  d["rain"].rolling(7).sum(),
        "rain_4w":  d["rain"].rolling(28).sum(),
        "rain_13w": d["rain"].rolling(91).sum(),
        "et0_4w":   d["et0"].rolling(28).sum(),
        "temp_1w":  d["temp"].rolling(7).mean(),
    })
    return f

def weather_variant(name):
    if name == "none":
        return None
    if name == "dam":
        return add_weather(wx_raw["dam"])
    if name == "upstream":                      # mean of the upstream points only
        return sum(add_weather(wx_raw[n]) for n in UPSTREAM) / len(UPSTREAM)
    if name == "dam+upstream":                  # dam + all upstream points averaged
        return sum(add_weather(wx_raw[n]) for n in points) / len(points)
    raise ValueError(name)

VARIANTS = ["none", "dam", "upstream", "dam+upstream"]
```

## 5. Features

```python
def make_features(weekly, wx=None, use_lag52=USE_LAG52):
    f = pd.DataFrame(index=weekly.index)
    f["pct_now"] = weekly["pct_obs"]                         # origin must be an actually-observed week
    for k in (1, 2, 4):
        f[f"d{k}w"] = weekly["pct_obs"] - weekly["pct_fill"].shift(k)    # lags may use causal single-gap fill
    if use_lag52:
        f["pct_52w_ago"] = weekly["pct_fill"].shift(52)
    if wx is not None:
        f = f.join(wx.reindex(weekly.index))
    woy = weekly.index.isocalendar().week.astype(int).values
    f["woy_sin"] = np.sin(2 * np.pi * woy / 52.18)
    f["woy_cos"] = np.cos(2 * np.pi * woy / 52.18)
    return f

def make_targets(weekly, horizons=HORIZONS):
    # target uses the interpolated series: it is the 'truth' we score against, never a feature
    return pd.DataFrame({h: weekly["pct_fill"].shift(-h) - weekly["pct_obs"] for h in horizons})

targets = make_targets(weekly)
feats = {v: make_features(weekly, weather_variant(v)) for v in VARIANTS}
feats["dam+upstream"].dropna().tail(3)
```

## 6. Time split (target dates never cross the cut-off)

```python
def split_time(features, targets, h, cutoff=CUTOFF):
    ok = features["pct_now"].notna() & targets[h].notna()
    X, y = features[ok], targets.loc[ok, h]
    train = X.index < cutoff - pd.Timedelta(weeks=h)
    test  = X.index >= cutoff
    return X[train], y[train], X[test], y[test]

for h in HORIZONS:
    Xtr, ytr, Xte, yte = split_time(feats["dam"], targets, h)
    print(f"h={h}w  train={len(Xtr)} ({Xtr.index.min().date()}..{Xtr.index.max().date()})  test={len(Xte)} ({Xte.index.min().date()}..{Xte.index.max().date()})")
```

## 7. Baselines (persistence and seasonal, computed from training data only)

```python
def seasonal_table(X_train, y_train):
    woy = X_train.index.isocalendar().week.astype(int)
    return y_train.groupby(woy.values).mean()

def seasonal_predict(table, index):
    woy = index.isocalendar().week.astype(int)
    return pd.Series(woy.map(table).fillna(0.0).values, index=index)

def season_mask(index):
    m = index.month.isin(MONSOON_MONTHS)
    return {"all": np.ones(len(index), bool), "monsoon": m, "dry": ~m}

def mae_by_season(y, pred, index):
    out = {}
    for k, m in season_mask(index).items():
        out[k] = float(np.mean(np.abs(np.asarray(y)[m] - np.asarray(pred)[m]))) if m.sum() else np.nan
    return out

def baselines(variant="dam"):
    rows = []
    for h in HORIZONS:
        Xtr, ytr, Xte, yte = split_time(feats[variant], targets, h)
        pers = mae_by_season(yte, np.zeros(len(yte)), Xte.index)
        seas = mae_by_season(yte, seasonal_predict(seasonal_table(Xtr, ytr), Xte.index), Xte.index)
        rows.append({"h_weeks": h, "n_test": len(yte),
                     **{f"persist_{k}": v for k, v in pers.items()}, **{f"seasonal_{k}": v for k, v in seas.items()}})
    return pd.DataFrame(rows).set_index("h_weeks").round(2)

baseline_table = baselines()
baseline_table
```

## 8. Models – HistGradientBoosting (median, p10, p90) with a Ridge reference
Grid chosen by `TimeSeriesSplit` MAE on training years only, with `gap=h`.

```python
def hgb(loss="absolute_error", quantile=None, **p):
    return HistGradientBoostingRegressor(loss=loss, quantile=quantile, max_iter=200, random_state=SEED, **p)

def cv_mae(model_fn, X, y, h, n_splits=5):
    errs = []
    for tr, va in TimeSeriesSplit(n_splits=n_splits, gap=h).split(X):
        m = model_fn().fit(X.iloc[tr], y.iloc[tr])
        errs.append(np.mean(np.abs(y.iloc[va] - m.predict(X.iloc[va]))))
    return float(np.mean(errs))

def tune_hgb(X, y, h):
    best = (np.inf, None)
    for md_, lr, leaf in itertools.product(GRID["max_depth"], GRID["learning_rate"], GRID["min_samples_leaf"]):
        p = dict(max_depth=md_, learning_rate=lr, min_samples_leaf=leaf)
        s = cv_mae(lambda: hgb(**p), X, y, h)
        if s < best[0]:
            best = (s, p)
    return best

def ridge_model():
    return make_pipeline(SimpleImputer(strategy="median"), StandardScaler(), Ridge(alpha=10.0))

def fit_quantiles(X, y, params):
    return {
        "q50": hgb(**params).fit(X, y),
        "q10": hgb(loss="quantile", quantile=0.10, **params).fit(X, y),
        "q90": hgb(loss="quantile", quantile=0.90, **params).fit(X, y),
    }

def predict_quantiles(models, X):
    q = np.column_stack([models["q10"].predict(X), models["q50"].predict(X), models["q90"].predict(X)])
    crossed = int((np.diff(q, axis=1) < 0).any(axis=1).sum())
    return np.sort(q, axis=1), crossed                       # enforce low <= median <= high
```

```python
# Tune every weather variant on training data only (CV), for every horizon
tuning = {}
t0 = time.time()
for v in VARIANTS:
    for h in HORIZONS:
        Xtr, ytr, _, _ = split_time(feats[v], targets, h)
        score, params = tune_hgb(Xtr, ytr, h)
        ridge_score = cv_mae(ridge_model, Xtr, ytr, h)
        tuning[(v, h)] = {"cv_mae": score, "params": params, "ridge_cv_mae": ridge_score}
print(f"tuning took {time.time() - t0:.0f}s")

cv_table = pd.DataFrame({(v, h): {"HGB cv MAE": t["cv_mae"], "Ridge cv MAE": t["ridge_cv_mae"]} for (v, h), t in tuning.items()}).T
cv_table.index.names = ["variant", "h_weeks"]
cv_table.round(3)
```

```python
# Weather variant is picked by mean CV MAE across horizons (training years only – not by looking at the test set)
variant_cv = {v: np.mean([tuning[(v, h)]["cv_mae"] for h in HORIZONS]) for v in VARIANTS}
BEST_VARIANT = min(variant_cv, key=variant_cv.get)
print({k: round(v, 3) for k, v in variant_cv.items()})
print("Chosen weather variant:", BEST_VARIANT)
```

## 9. Evaluation on the test period (2022 →)

```python
def evaluate_variant(v):
    res = {}
    for h in HORIZONS:
        Xtr, ytr, Xte, yte = split_time(feats[v], targets, h)
        models = fit_quantiles(Xtr, ytr, tuning[(v, h)]["params"])
        q, crossed = predict_quantiles(models, Xte)
        seas_tab = seasonal_table(*split_time(feats["dam"], targets, h)[:2])
        seas = seasonal_predict(seas_tab, Xte.index)
        ridge = ridge_model().fit(Xtr, ytr)
        res[h] = dict(
            models=models, X_test=Xte, y_test=yte, q=q, crossed=crossed, seasonal=seas,
            model_mae=mae_by_season(yte, q[:, 1], Xte.index),
            seasonal_mae=mae_by_season(yte, seas, Xte.index),
            persist_mae=mae_by_season(yte, np.zeros(len(yte)), Xte.index),
            ridge_mae=mae_by_season(yte, ridge.predict(Xte), Xte.index),
            coverage=float(((yte.values >= q[:, 0]) & (yte.values <= q[:, 2])).mean()),
        )
    return res

results = {v: evaluate_variant(v) for v in VARIANTS}

rows = []
for v in VARIANTS:
    for h in HORIZONS:
        r = results[v][h]
        rows.append({"variant": v, "h_weeks": h,
                     "MAE_all": r["model_mae"]["all"], "MAE_monsoon": r["model_mae"]["monsoon"], "MAE_dry": r["model_mae"]["dry"],
                     "seasonal_all": r["seasonal_mae"]["all"], "seasonal_monsoon": r["seasonal_mae"]["monsoon"], "seasonal_dry": r["seasonal_mae"]["dry"],
                     "persist_all": r["persist_mae"]["all"],
                     "ridge_all": r["ridge_mae"]["all"], "ridge_monsoon": r["ridge_mae"]["monsoon"],
                     "skill_all": 1 - r["model_mae"]["all"] / r["seasonal_mae"]["all"],
                     "skill_monsoon": 1 - r["model_mae"]["monsoon"] / r["seasonal_mae"]["monsoon"],
                     "skill_dry": 1 - r["model_mae"]["dry"] / r["seasonal_mae"]["dry"],
                     "band_coverage": r["coverage"], "quantile_crossings_fixed": r["crossed"]})
eval_table = pd.DataFrame(rows).set_index(["variant", "h_weeks"])
eval_table.round(3)
```

```python
# Does upstream weather help in the monsoon? (the comparison worth putting in the write-up)
monsoon_cmp = eval_table["MAE_monsoon"].unstack("variant").round(2)
monsoon_cmp["seasonal"] = eval_table.xs("dam", level="variant")["seasonal_monsoon"].round(2)
print("Monsoon MAE (pct-points) by weather variant:")
display(monsoon_cmp)
for h in HORIZONS:
    s = eval_table.loc[(BEST_VARIANT, h), "skill_monsoon"]
    if s <= 0:
        print(f"NOTE: h={h}w monsoon skill vs seasonal baseline is {s:.2f} (<= 0) – the model does not beat the seasonal baseline there.")
```

## 10. Low-storage warning
Threshold = 20th percentile of **training** `pct_full`. Only **new crossings** are scored (storage above the threshold now, below it at t+h). A warning fires when the predicted level `pct_now + prediction` is below the threshold — using the median (`q50`) or the cautious low band (`q10`).

```python
def warning_scores(origin_pct, actual_future_pct, pred_future_pct, thr):
    new = origin_pct >= thr
    event = new & (actual_future_pct < thr)
    warn = new & (pred_future_pct < thr)
    tp = int((event & warn).sum()); fn = int((event & ~warn).sum()); fp = int((~event & warn & new).sum())
    return dict(events=int(event.sum()), warned=int(warn.sum()), hits=tp, missed=fn, false_alarms=fp,
                recall=tp / (tp + fn) if tp + fn else np.nan, precision=tp / (tp + fp) if tp + fp else np.nan)

THRESHOLD = float(np.percentile(clean.loc[clean.date < CUTOFF, "pct_full"].dropna(), WARN_PERCENTILE))
print(f"Warning threshold: {THRESHOLD:.1f} % full")

warn_rows = []
for h in HORIZONS:
    r = results[BEST_VARIANT][h]
    origin = r["X_test"]["pct_now"].values
    actual = origin + r["y_test"].values
    for name, col in (("median", 1), ("p10_band", 0)):
        warn_rows.append({"h_weeks": h, "rule": name, **warning_scores(origin, actual, origin + r["q"][:, col], THRESHOLD)})
warn_test = pd.DataFrame(warn_rows).set_index(["h_weeks", "rule"])
warn_test.round(2)
```

```python
# Rolling backtest: for each year Y (2016-2025) train on targets that end before Y, predict Y. Pools many more events.
def threshold_for(train_end):
    return float(np.percentile(clean.loc[clean.date < train_end, "pct_full"].dropna(), WARN_PERCENTILE))

bt = []
for h in HORIZONS:
    X_all, y_all = feats[BEST_VARIANT], targets[h]
    ok = X_all["pct_now"].notna() & y_all.notna()
    X_all, y_all = X_all[ok], y_all[ok]
    params = tuning[(BEST_VARIANT, h)]["params"]
    for Y in BACKTEST_YEARS:
        start = pd.Timestamp(f"{Y}-01-01"); end = pd.Timestamp(f"{Y + 1}-01-01")
        tr = X_all.index < start - pd.Timedelta(weeks=h)
        te = (X_all.index >= start) & (X_all.index < end)
        if te.sum() == 0 or tr.sum() < 100:
            continue
        models = fit_quantiles(X_all[tr], y_all[tr], params)
        q, _ = predict_quantiles(models, X_all[te])
        seas = seasonal_predict(seasonal_table(X_all[tr], y_all[tr]), X_all[te].index)
        bt.append(pd.DataFrame({
            "date": X_all[te].index, "h_weeks": h, "fold_year": Y, "threshold": threshold_for(start),
            "pct_now": X_all[te]["pct_now"].values, "actual_change": y_all[te].values,
            "pred_lo": q[:, 0], "pred_med": q[:, 1], "pred_hi": q[:, 2],
            "pred_seasonal": seas.values, "pred_persist": 0.0}))
backtest = pd.concat(bt, ignore_index=True)
backtest.to_csv(OUT_DIR / "backtest.csv", index=False)

bt_rows = []
for h, g in backtest.groupby("h_weeks"):
    for name, col in (("median", "pred_med"), ("p10_band", "pred_lo"), ("seasonal", "pred_seasonal")):
        bt_rows.append({"h_weeks": h, "rule": name, **warning_scores(
            g.pct_now.values, (g.pct_now + g.actual_change).values, (g.pct_now + g[col]).values, g.threshold.values)})
warn_backtest = pd.DataFrame(bt_rows).set_index(["h_weeks", "rule"])
print("Rolling backtest pooled over", backtest.fold_year.min(), "-", backtest.fold_year.max())
warn_backtest.round(2)
```

```python
# Backtest MAE by season, per horizon
def season_of(d): return np.where(pd.DatetimeIndex(d).month.isin(MONSOON_MONTHS), "monsoon", "dry")
b = backtest.assign(season=season_of(backtest.date))
for c in ("pred_med", "pred_seasonal", "pred_persist"):
    b[f"err_{c}"] = (b.actual_change - b[c]).abs()
b.groupby(["h_weeks", "season"])[["err_pred_med", "err_pred_seasonal", "err_pred_persist"]].mean().round(2)
```

## 11. Final fit on all data and save

```python
FINAL = {}
for h in HORIZONS:
    X_all, y_all = feats[BEST_VARIANT], targets[h]
    ok = X_all["pct_now"].notna() & y_all.notna()
    X_all, y_all = X_all[ok], y_all[ok]
    FINAL[h] = {"models": fit_quantiles(X_all, y_all, tuning[(BEST_VARIANT, h)]["params"]),
                "seasonal_table": seasonal_table(X_all, y_all).to_dict(),
                "params": tuning[(BEST_VARIANT, h)]["params"]}

FEATURE_COLS = list(feats[BEST_VARIANT].columns)
bundle = {
    "version": VERSION, "horizons_weeks": list(HORIZONS), "feature_cols": FEATURE_COLS,
    "weather_variant": BEST_VARIANT, "upstream_points": UPSTREAM, "dam_coords": [DAM_LAT, DAM_LON],
    "warning_threshold_pct": THRESHOLD, "implied_capacity_bcm": float(IMPLIED_CAPACITY),
    "data_end_date": str(clean.date.max().date()), "trained_on": "all data", "models": FINAL,
}
joblib.dump(bundle, OUT_DIR / "models.joblib")

# latest forecast: the latest week that has an observed pct_full AND all features
Xf = feats[BEST_VARIANT]
latest = Xf.dropna().index.max()
rows = []
for h in HORIZONS:
    q, _ = predict_quantiles(FINAL[h]["models"], Xf.loc[[latest], FEATURE_COLS])
    now = float(Xf.loc[latest, "pct_now"])
    rows.append({"origin_week_end": latest.date(), "horizon_weeks": h, "target_week_end": (latest + pd.Timedelta(weeks=h)).date(),
                 "pct_full_now": round(now, 2), "pred_low": round(now + q[0, 0], 2), "pred_median": round(now + q[0, 1], 2),
                 "pred_high": round(now + q[0, 2], 2), "below_threshold_median": bool(now + q[0, 1] < THRESHOLD)})
latest_forecast = pd.DataFrame(rows)
latest_forecast.to_csv(OUT_DIR / "latest_forecast.csv", index=False)

metrics = {
    "version": VERSION, "weather_variant": BEST_VARIANT, "cutoff": str(CUTOFF.date()), "warning_threshold_pct": THRESHOLD,
    "units": "percentage points of pct_full",
    "variant_cv_mae": variant_cv,
    "test": {f"{h}w": {k: (float(v) if not isinstance(v, dict) else {a: float(b_) for a, b_ in v.items()})
                       for k, v in results[BEST_VARIANT][h].items() if k in ("model_mae", "seasonal_mae", "persist_mae", "ridge_mae", "coverage")}
             for h in HORIZONS},
    "test_skill_vs_seasonal": {f"{h}w": {k: float(eval_table.loc[(BEST_VARIANT, h), f"skill_{k}"]) for k in ("all", "monsoon", "dry")} for h in HORIZONS},
    "monsoon_mae_by_variant": {v: {f"{h}w": float(eval_table.loc[(v, h), "MAE_monsoon"]) for h in HORIZONS} for v in VARIANTS},
    "warning_test": json.loads(warn_test.reset_index().to_json(orient="records")),
    "warning_backtest": json.loads(warn_backtest.reset_index().to_json(orient="records")),
}
json.dump(metrics, open(OUT_DIR / "metrics.json", "w"), indent=2, default=float)
print("saved:", sorted(p.name for p in OUT_DIR.iterdir()))
latest_forecast
```

## 12. Sanity checks

```python
# (a) No feature may use data after its own date: rebuild features from data truncated at t and compare
rng = np.random.default_rng(SEED)
full = feats[BEST_VARIANT]
candidates = full.dropna().index[60:]
samples = rng.choice(candidates, size=25, replace=False)
for t in samples:
    t = pd.Timestamp(t)
    w_t = to_weekly(clean[clean.date <= t])
    wx_t = {n: d[d.index <= t] for n, d in wx_raw.items()}
    wx_variant = None
    if BEST_VARIANT == "dam":
        wx_variant = add_weather(wx_t["dam"])
    elif BEST_VARIANT == "upstream":
        wx_variant = sum(add_weather(wx_t[n]) for n in UPSTREAM) / len(UPSTREAM)
    elif BEST_VARIANT == "dam+upstream":
        wx_variant = sum(add_weather(wx_t[n]) for n in points) / len(points)
    f_t = make_features(w_t, wx_variant).loc[t]
    pd.testing.assert_series_equal(f_t[FEATURE_COLS], full.loc[t, FEATURE_COLS], check_names=False, atol=1e-9)
print("OK: features at each of 25 sampled weeks are identical when all later data is removed (no look-ahead).")

# (b) quantile ordering
for h in HORIZONS:
    q = results[BEST_VARIANT][h]["q"]
    assert (q[:, 0] <= q[:, 1]).all() and (q[:, 1] <= q[:, 2]).all()
print("OK: low <= median <= high for every test prediction.")

# (c) train targets never reach the test period
for h in HORIZONS:
    Xtr, ytr, Xte, yte = split_time(feats[BEST_VARIANT], targets, h)
    assert (Xtr.index + pd.Timedelta(weeks=h) < CUTOFF).all() and Xte.index.min() >= CUTOFF
print("OK: no training target date falls on/after the cut-off.")
```

```python
# (d) predicted vs actual (test period)
fig, axes = plt.subplots(len(HORIZONS), 2, figsize=(14, 3.6 * len(HORIZONS)))
for i, h in enumerate(HORIZONS):
    r = results[BEST_VARIANT][h]
    idx, y, q = r["X_test"].index, r["y_test"].values, r["q"]
    ax = axes[i, 0]
    ax.fill_between(idx, q[:, 0], q[:, 2], alpha=.25, label="p10-p90")
    ax.plot(idx, y, "k.", ms=3, label="actual"); ax.plot(idx, q[:, 1], "r-", lw=1, label="median")
    ax.set_title(f"{h}-week change (pct points)"); ax.legend(fontsize=7)
    ax = axes[i, 1]
    ax.scatter(y, q[:, 1], s=6); lim = [min(y.min(), q[:, 1].min()), max(y.max(), q[:, 1].max())]
    ax.plot(lim, lim, "k--", lw=.8); ax.set_xlabel("actual"); ax.set_ylabel("predicted"); ax.set_title(f"{h}-week: predicted vs actual")
plt.tight_layout(); plt.savefig(OUT_DIR / "pred_vs_actual.png", dpi=120); plt.show()
```

## Notes for the write-up
- Call the longest horizon **4 weeks** (28 days), not "30 days".
- Hirakud monsoon changes depend on gate-release decisions the model cannot see – report monsoon skill honestly, including if it is ≤ 0.
- Pre-2013 `storage_bcm` was re-derived from level; only `pct_full` (through the current curve) is used here, with `implied_capacity_bcm` stored in the model bundle for days-of-cover.
